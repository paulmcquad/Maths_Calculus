```python
import sympy as sp

# 1. Define your symbolic variable
x = sp.symbols('x')

# 2. Standard Finite Limit
# Evaluates: lim (x -> 0) of sin(x) / x
expr1 = sp.sin(x) / x
limit_finite = sp.limit(expr1, x, 0)
print(f"1. Finite Limit: {limit_finite}")  # Output: 1

# 3. Limit at Infinity (Use sp.oo for positive infinity, -sp.oo for negative)
# Evaluates: lim (x -> oo) of 1 / x
expr2 = 1 / x
limit_inf = sp.limit(expr2, x, sp.oo)
print(f"2. Limit at Infinity: {limit_inf}")  # Output: 0

# 4. One-Sided Limits (Use dir='+' or dir='-')
# Right-hand limit: lim (x -> 0+) of 1 / x
limit_right = sp.limit(expr2, x, 0, dir='+') 
# Left-hand limit: lim (x -> 0-) of 1 / x
limit_left = sp.limit(expr2, x, 0, dir='-')   

print(f"3. Right-hand limit (0+): {limit_right}")  # Output: oo
print(f"4. Left-hand limit (0-): {limit_left}")    # Output: -oo

# 5. Unevaluated Limit (Useful for pretty-printing or latex generation)
# Creates a representation without immediately calculating it
lazy_limit = sp.Limit(sp.cos(x), x, sp.pi)
print(f"5. Unevaluated Object: {lazy_limit}")  # Output: Limit(cos(x), x, pi)

# Evaluate the unevaluated object using .doit()
evaluated_lazy = lazy_limit.doit()
print(f"6. Evaluated Object: {evaluated_lazy}")  # Output: -1
```
