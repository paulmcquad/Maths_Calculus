```python
import sympy as sp

x = sp.symbols('x')
# Define a piecewise function
f = sp.Piecewise((x, x != 2), (3, x == 2))

# Compute limit as x approaches 1 from both sides

lim_val = sp.limit(f, x, 2)
lim_val
```

```python
ls  = sp.limit(sp.sin(x)/x, x, 0)
ls
```
