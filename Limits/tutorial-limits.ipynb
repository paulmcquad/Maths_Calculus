```python
import matplotlib.pyplot as plt
import numpy as np

# Define the function
def f(x):
    return (x**3 - 8) / (x**2 - 4)

# Create x values in two ranges to avoid the vertical asymptote at x = -2
x1 = np.linspace(-10, -2.01, 400)
x2 = np.linspace(-1.99, 10, 400)

# Create the plot
plt.figure(figsize=(8, 6))

# Plot the two branches of the function
plt.plot(x1, f(x1), "b-", label=r"$f(x) = \frac{x^3 - 8}{x^2 - 4}$")
plt.plot(x2, f(x2), "b-")

# Plot the slant asymptote y = x
x_asymptote = np.linspace(-10, 10, 100)
plt.plot(x_asymptote, x_asymptote, "r--", label="Slant Asymptote: $y = x$")

# Plot the vertical asymptote at x = -2
plt.axvline(x=-2, color="g", linestyle="--", label="Vertical Asymptote: $x = -2$")

# Mark the removable discontinuity (hole) at (2, 3)
plt.plot(2, 3, "ro", markerfacecolor="white", markersize=8, label="Hole at (2, 3)")

# Formatting
plt.xlim(-10, 10)
plt.ylim(-10, 10)
plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.grid(True, linestyle=":", alpha=0.6)
plt.title("Plot of rational function with asymptotes and hole")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Define x values strictly greater than -1, avoiding exactly -1
x = np.linspace(-0.99, 4.0, 1000)

# 2. Safely calculate the function using piecewise logic to handle x = 0
f = np.piecewise(
    x,
    [x == 0, x != 0],
    [lambda x: np.e, lambda x: (1 + x) ** (1 / x)],
)

# 3. Create the plot
plt.figure(figsize=(8, 5))
plt.plot(x, f, label=r"$f(x) = (1+x)^{1/x}$", color="blue", linewidth=2)

# 4. Highlight the limit point (0, e)
plt.scatter([0], [np.e], color="red", facecolors="none", s=80, zorder=5)
plt.text(
    0.15,
    np.e,
    f"(0, e) \ne ≈ {np.e:.4f}",
    color="red",
    fontsize=10,
    va="center",
)

# 5. Labels and styling
plt.title(r"Plot of $f(x) = (1+x)^{1/x}$", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("f(x)", fontsize=12)
plt.axhline(1, color="gray", linestyle="--", alpha=0.5, label="y = 1")
plt.axvline(0, color="gray", linestyle="--", alpha=0.5)
plt.grid(True, alpha=0.3)
plt.legend()

# 6. Show the graph
plt.show()
```
